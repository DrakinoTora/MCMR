```python
import mcmr
```

```python
N = 10000
material = "Fe"
x1 = 15
y1 = 10
x2 = 35
y2 = 35
x_world = 50
y_world = 50
```

```python
sim = mcmr.load_and_run(
    N=N, 
    material_name=material, 
    box_coords=(x1, y1, x2, y2), 
    world_coords=(x_world, y_world)
)
```

```python
plotter = mcmr.ResultsPlotter("mcmr_results.xml")
```

```python
plotter.plot_trajectories(box_bounds=(15.0, 10.0, 35.0, 35.0), world_bounds=(50.0, 50.0))
```

```python
plotter.plot_energy_born()
```
