```python
import mcmr
```

```python
N = 10000
x = 50
y = 50
x_grid=[3, 15, 21]
y_grid=[10, 20, 41]

material=[
    ["Fe", "Pb", "C",  "Fe"],
    ["Pb", "Fe", "Pb", "C"],
    ["C",  "Pb", "Fe", "Pb"],
    ["Fe", "C",  "Pb", "Fe"],
]
```

```python
sim = mcmr.load_and_run(
    N=N,
    x_world=x, y_world=y,
    x_grid=x_grid,
    y_grid=x_grid,
    material_matrix=material,
)
```

```python
t = sim.get_tally()
```

```python
print(t.absorp_by_material)
```

```python
plotter = mcmr.ResultsPlotter("mcmr_results.xml")
```

```python
plotter.plot_trajectories(box_bounds=(15.0, 10.0, 35.0, 35.0), world_bounds=(50.0, 50.0))
```
