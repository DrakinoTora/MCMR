```python
import mcmr

# 1. Panggil parser HTML (Python) langsung dari mcmr
E_Al_t, Sig_Al_t = mcmr.load_cross_section('2631', 1)
E_Al_s, Sig_Al_s = mcmr.load_cross_section('2631', 2)
E_Pb_t, Sig_Pb_t = mcmr.load_cross_section('8237', 1)
E_Pb_s, Sig_Pb_s = mcmr.load_cross_section('8237', 2)

# 2. Panggil engine simulasi (C++) langsung dari mcmr
res = mcmr.neutron_sim(
    15.0, 45.0, 45.0, 1000, False,
    E_Al_t, Sig_Al_t, E_Al_s, Sig_Al_s,
    E_Pb_t, Sig_Pb_t, E_Pb_s, Sig_Pb_s
)

print(f"Transmisi: {res.transmisi}")
print(f"Waktu C++: {res.time_taken:.4f} detik")
```
