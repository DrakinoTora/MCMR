```python
import mcmr
```

```python
sim = mcmr.load_and_run(
    N=100000, 
    material_name="Fe", 
    box_coords=(15.0, 10.0, 35.0, 35.0), 
    world_coords=(50.0, 50.0)
)
```

```python
plotter = mcmr.ResultsPlotter("mcmr_results.xml")
```

```python
plotter.plot_trajectories(box_bounds=(15.0, 10.0, 35.0, 35.0), world_bounds=(50.0, 50.0))
```

```python
plotter.plot_energy_born()
```

```python
plotter.plot_energy_leak()
```
